```python
from pathlib import Path
import os
import subprocess
import sys

# ===== Change these values only =====
PROMPT = 'a wooden chair, studio product photograph, centered object, white background'
OUTPUT_DIR = '/kaggle/working/chair-output'
SEED = '42'

REPO_URL = 'https://github.com/huytrao/carve3d-test.git'
BRANCH = 'implement_full_pipeline'
REPO_DIR = Path('/kaggle/working/carve3d-test')
LGM_ROOT = '/kaggle/working/LGM'

if not (REPO_DIR / '.git').is_dir():
    subprocess.run(['git', 'clone', '--depth', '1', '--branch', BRANCH, REPO_URL, str(REPO_DIR)], check=True)
else:
    subprocess.run(['git', '-C', str(REPO_DIR), 'fetch', '--depth', '1', 'origin', BRANCH], check=True)
    subprocess.run(['git', '-C', str(REPO_DIR), 'checkout', '-B', BRANCH, f'origin/{BRANCH}'], check=True)

# LGM 2024 requires this kiui release. Force reinstall before importing any LGM/MVDream code.
subprocess.run([sys.executable, '-m', 'pip', 'install', '--no-deps', '--force-reinstall', 'kiui==0.2.3'], check=True)

# Force this cell to use the code just fetched, not modules cached by a previous failed run.
for module_name in list(sys.modules):
    if module_name in {'kaggle_full_pipeline_prompt', 'run_full_pipeline'} or module_name.startswith(('full_pipeline', 'mvdream', 'core', 'kiui')):
        sys.modules.pop(module_name, None)
import importlib
importlib.invalidate_caches()
sys.path.insert(0, str(REPO_DIR))
from kaggle_full_pipeline_prompt import main as run_prompt_pipeline

# GPU 0: MVDream. GPU 1: LGM reconstruction, rendering, MRC.
run_prompt_pipeline([
    '--lgm-root', LGM_ROOT,
    '--prompt', PROMPT,
    '--seed', SEED,
    '--output-dir', OUTPUT_DIR,
])
```
